# Decision Tree Rules Extraction & Visualization

Notebook phân tích khả năng giải thích (Explanability), trích xuất các quy tắc (Decision Rules), tầm quan trọng đặc trưng (Feature Importance) và trực quan hóa cây quyết định tối ưu cho báo cáo, slide và video.

```python
import json
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier, plot_tree

sns.set_theme(style="whitegrid")
```

## 1. Nạp dữ liệu và Huấn luyện Mô hình

```python
DATA_PATH = Path("../../data/processed/letter_recognition/train.csv")
TEST_PATH = Path("../../data/processed/letter_recognition/test.csv")

train_df = pd.read_csv(DATA_PATH)
test_df = pd.read_csv(TEST_PATH)

FEATURES = [c for c in train_df.columns if c != "letter"]
TARGET = "letter"

X_train, y_train = train_df[FEATURES], train_df[TARGET]
X_test, y_test = test_df[FEATURES], test_df[TARGET]

clf = DecisionTreeClassifier(criterion="gini", random_state=42)
clf.fit(X_train, y_train)

print(f"Tree Depth: {clf.get_depth()}, Leaves: {clf.get_n_leaves()}")
```

## 2. Trực quan hóa Feature Importance

```python
imp = pd.Series(clf.feature_importances_, index=FEATURES).sort_values(ascending=False)
plt.figure(figsize=(10, 6))
sns.barplot(x=imp.values, y=imp.index, palette="viridis")
plt.title("Decision Tree Feature Importance - Letter Recognition")
plt.xlabel("Gini Importance")
plt.show()
```

## 3. Trực quan hóa Cây Quyết định 3 Tầng Đầu

```python
plt.figure(figsize=(24, 12), dpi=200)
plot_tree(clf, feature_names=FEATURES, class_names=clf.classes_, max_depth=3, filled=True, rounded=True, fontsize=10)
plt.title("First 3 Levels of Best Decision Tree")
plt.show()
```
